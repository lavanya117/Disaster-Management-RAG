```python
def scraper(url):
    driver=webdriver.Chrome(service=service)
    try:
        links=[]
        driver.get(url)
        anch_link=driver.find_elements(By.TAG_NAME,'a')
        for link in anch_link:
            urls_all=link.get_attribute('href')
            links.append(urls_all)
        return links     
    
    finally:
        driver.quit()
```

```python
all_links=[]
```

```python
for url in tqdm(urls_list):
    links=scraper(url)
    for link in links:
        all_links.append(link)
```

```python
#remove duplicates
from urllib.parse import urlparse

def norm(url):
    p=urlparse(url.lower().strip())
    netloc=p.netloc.replace('www',' ')
    path=p.path.rstrip('/')
    return f"{netloc}{path}"
```

```python
seen={}
for url in new_links:
    key=norm(url)
    if key not in seen:
        seen[key]=url
    
unique_urls=list(seen.values())    
   
```
